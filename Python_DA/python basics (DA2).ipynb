# While loop in python

```python
nu=0
while nu<=10:
    print(nu)
    nu=nu+1
```

```python
nu=0
while nu<=10:
    print(nu)
    if nu==5:
        break
    nu=nu+1
```

```python
nu=0
while nu<5:
    print(nu)
    if nu==10:
        break
    nu=nu+1
else:
    print("Not less than 10")
```

```python
nu=0
while nu<5:
    nu=nu+1
    
    if nu==3:
        continue
    print(nu)
else:
    print("Not less than 10")
```

# BMI calculator

```python
print(3.28*3.28)
```

```python
# BMI= weight in kg /height2 in m

weight=float(input("Please Enter your weight in KG: "))
height=float(input("Please Enter your height in foot: "))
name=input("Please Enter your name: ")
# height=height*3.28
BMI=(weight*10.7583)/(height*height)
if BMI>0:
    if BMI<18.5:
        print(f"Hi {name}!, Your BMI is {BMI} and you fall into Underweight category.")
    elif BMI>=18.5 and BMI<=24.9:
        print(f"Hi {name}!, Your BMI is {BMI} and you fall into N0rmal Weight category.")
    elif BMI>=25 and BMI<=29.9:
        print(f"Hi {name}!, Your BMI is {BMI} and you fall into OverWeight category.")
    elif BMI>=30 and BMI<=35:
        print(f"Hi {name}!, Your BMI is {BMI} and you fall into Obese category.")
    elif BMI>35 and BMI<40:
        print(f"Hi {name}!, Your BMI is {BMI} and you fall into Morbid obesity category.")
    else:
        print(f"Hi {name}!, Your BMI is {BMI} and you fall into severe Morbid obesity category.")
else:
    print("Please Enter valid values")

    
# print(f"Your Body Mass Index is: ",{BMI})
```

# Automatic File Sorter in File Explorer

```python
import os, shutil
```

```python
path=r"C:/DATA ANALYST/python_DA/"
```

```python
os.path.exists(path)
```

```python
file_name=os.listdir(path)
```

```python
folder_names=['csv files', 'txt files', 'image files', 'pdf files']
for loop in range(0,4):
    if not os.path.exists(path + folder_names[loop]):
        print(path + folder_names[loop])
        os.makedirs(path + folder_names[loop])
    else:
        print("Folder already present.")

for file in file_name:
    if '.csv' in file and not os.path.exists(path+'csv files/'+ file):
        shutil.move(path+file,path+'csv files/'+ file)
    elif '.txt' in file and not os.path.exists(path+'txt files/'+ file):
        shutil.move(path+file,path+'txt files/'+ file)
    elif '.png' in file and not os.path.exists(path+'image files/'+ file):
        shutil.move(path+file,path+'image files/'+ file)
    elif '.jpeg' in file and not os.path.exists(path+'image files/'+ file):
        shutil.move(path+file,path+'image files/'+ file)
    elif '.pdf' in file and not os.path.exists(path+'pdf files/'+ file):
        shutil.move(path+file,path+'pdf files/'+ file)
    
        

#os.path.exists(path+'csv files')
```
