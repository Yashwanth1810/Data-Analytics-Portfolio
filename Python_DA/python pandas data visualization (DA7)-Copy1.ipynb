# visualization with pandas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\DATA ANALYST\python_DA\original_practice_python\Ice Cream Ratings.csv")
df=df.set_index('Date')
df
```

```python
print(plt.style.available)
plt.style.use('fivethirtyeight')
```

```python
df.plot(kind='line',title='Ice Cream Ratings',legend='Overall Rating',ylabel="Overall Rating",xlabel='Monthly Ratings',figsize=(9,6))
```

```python
df.plot(kind='bar',ylabel="Overall Rating",xlabel='Monthly Ratings',stacked=True)
```

```python
df['Flavor Rating'].plot(kind='bar')
```

```python
df.plot.barh(stacked=True)
```

```python
df.plot.scatter(y='Overall Rating',x='Flavor Rating',s=500,c='green')
```

```python
df.plot.hist(bins=20)
```

```python
df.boxplot()
```

```python
df.plot.area(figsize=(10,5))
```

```python
df.plot.pie(y='Flavor Rating',figsize=(10,6))
```
