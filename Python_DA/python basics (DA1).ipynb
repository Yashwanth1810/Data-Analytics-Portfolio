# Logical Operators

### AND          Returns TRUE only if both the values are TRUE else FALSE
### OR           Returns TRUE if any of the values are TRUE, if both the values are FALSE then FALSE
### NOT          if TRUE then FALSE, if FALSE then TRUE(Reverses the result)



```python
(10>2) and (2<30)
```

```python
(10>2) and (200<30)
```

```python
(10>2) or (200<30)
```

```python
not(100>20)
```

```python
dic={"Data Analyst":['SQL',"python", "powerBI","Excel"], "DAta Scientist": "Machine learning models", "Experience":5}
for i in dic.items():
    print(i, end=" ")
    print()
```

```python
dic={"Data Analyst":['SQL',"python", "powerBI","Excel"], "DAta Scientist": "Machine learning models", "Experience":5}
for key, value in dic.items():
    print(key,'->',value)
```

```python
# Nested for loops

Data_Analyst=['SQL',"python", "powerBI","Excel"]
Data_Scientist=["ML Models", "ML Algorithms", "Cloud Services"]
for one in Data_Analyst:
    for two in Data_Scientist:
        print(one ,"shifts to", two)
```
